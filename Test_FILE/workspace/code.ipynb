```
from sympy import gcd, lcm

# Define the known values
product = 120  # This is the product of m and n, i.e., m * n

# Possible pairs (m, n) that satisfy the condition
possible_pairs = []
for m in range(1, int(product**0.5) + 1):
    if product % m == 0:  # m must be a divisor of the product
        n = product // m
        if gcd(m, n) * lcm(m, n) == product:  # Check the relationship
            possible_pairs.append((m, n))

possible_pairs
```

```
min_sum = min(m + n for m, n in possible_pairs)
min_sum
```
