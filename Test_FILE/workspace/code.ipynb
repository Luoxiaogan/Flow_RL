```
from math import gcd

# Given LCM and product of m and n
lcm_value = 120
product_mn = 3600

# Calculate GCD using the relationship GCD(m, n) * LCM(m, n) = m * n
gcd_value = product_mn // lcm_value

# Possible pairs (m, n) that satisfy the conditions
possible_pairs = []
for m in range(1, int(product_mn**0.5) + 1):
    if product_mn % m == 0:  # m must be a divisor of the product
        n = product_mn // m
        if gcd(m, n) == gcd_value:  # Check if the GCD condition is satisfied
            possible_pairs.append((m, n))

possible_pairs
```

```
min_sum = min(m + n for m, n in possible_pairs)
min_sum
```
