```python
file_path = './0722_new_mbpp/training_data_mbpp.jsonl'

print("对于MBPP数据")
length_list = []
# Read the file line by line
with open(file_path, 'r') as file:
    for line in file:
        length_list.append(len(line.strip()))  # Store the length of each line after stripping whitespace
        # print(len(line.strip()))  # Print the length of each line after stripping whitespace
print("max length:", max(length_list))
print("min length:", min(length_list))
# Analyze the lengths
if length_list:
    average_length = sum(length_list) / len(length_list)
    print(f"Average length of lines: {average_length}")
```
